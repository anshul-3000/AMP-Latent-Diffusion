```python
#properties extraction for clean_amp dataset

import pandas as pd

amp_df = pd.read_csv(
    "../data/processed_data&code/clean_amp_dataset.csv"
)

print(amp_df.shape)
```

```python
#resuable function for extracting properties

from Bio.SeqUtils.ProtParam import ProteinAnalysis
import pandas as pd
from tqdm import tqdm

def net_charge(seq):

    positive = seq.count("K") + seq.count("R")
    negative = seq.count("D") + seq.count("E")

    return positive - negative


def extract_properties(df, sequence_col="Sequence"):

    properties = []

    for seq in tqdm(df[sequence_col]):

        try:

            analysis = ProteinAnalysis(seq)

            properties.append({

                "Length": len(seq),

                "MolecularWeight":
                    analysis.molecular_weight(),

                "Aromaticity":
                    analysis.aromaticity(),

                "InstabilityIndex":
                    analysis.instability_index(),

                "IsoelectricPoint":
                    analysis.isoelectric_point(),

                "Hydrophobicity":
                    analysis.gravy(),

                "NetCharge":
                    net_charge(seq)

            })

        except:

            properties.append({

                "Length": None,
                "MolecularWeight": None,
                "Aromaticity": None,
                "InstabilityIndex": None,
                "IsoelectricPoint": None,
                "Hydrophobicity": None,
                "NetCharge": None

            })

    return pd.DataFrame(properties)
```

```python
amp_df = pd.read_csv(
    "../data/processed_data&code/clean_amp_dataset.csv"
)

amp_properties = extract_properties(amp_df)

print(amp_properties.shape)

amp_properties.to_csv(
    "amp_properties.csv",
    index=False
)

print("AMP properties saved")
```
