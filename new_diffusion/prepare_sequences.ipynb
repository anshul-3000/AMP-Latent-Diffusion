```python
import pandas as pd
import numpy as np

amp_df = pd.read_csv(
    "../data/processed_data&code/clean_amp_dataset.csv"
)

print(amp_df.shape)

print(
    amp_df["Sequence"]
    .str.len()
    .describe()
)
```

```python
AA_VOCAB = "ACDEFGHIKLMNPQRSTVWY"

aa_to_idx = {
    aa: i + 1
    for i, aa in enumerate(AA_VOCAB)
}

aa_to_idx["PAD"] = 0

MAX_LEN = 60

encoded_sequences = []

for seq in amp_df["Sequence"]:

    tokens = [
        aa_to_idx[aa]
        for aa in seq
    ]

    tokens += [0] * (
        MAX_LEN - len(tokens)
    )

    encoded_sequences.append(tokens)

encoded_sequences = np.array(
    encoded_sequences
)

print(encoded_sequences.shape)
print(encoded_sequences.dtype)
print(encoded_sequences.max())
print(encoded_sequences.min())
```
