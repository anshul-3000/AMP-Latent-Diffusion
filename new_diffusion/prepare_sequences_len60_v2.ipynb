```python
import pandas as pd

amp_df = pd.read_csv(
    "../data/processed_data&code/clean_amp_dataset.csv"
)

print(amp_df.shape)
```

```python
amp_df["Length"] = (
    amp_df["Sequence"]
    .str.len()
)

print(
    amp_df["Length"]
    .describe()
)
```

```python
#Create >40 dataset.
long_df = amp_df[
    (amp_df["Length"] >= 41)
    &
    (amp_df["Length"] <= 60)
].copy()

print(long_df.shape)
```

```python
PAD = 0
START = 1
END = 2

AA_VOCAB = "ACDEFGHIKLMNPQRSTVWY"

aa_to_idx = {
    aa:i+3
    for i,aa
    in enumerate(AA_VOCAB)
}
```

```python
import numpy as np
MAX_LEN = 62

encoded_sequences = []

for seq in long_df["Sequence"]:

    tokens = [START]

    tokens += [
        aa_to_idx[aa]
        for aa in seq
    ]

    tokens.append(END)

    tokens += [PAD] * (
        MAX_LEN - len(tokens)
    )

    encoded_sequences.append(tokens)

encoded_sequences = np.array(
    encoded_sequences
)

print(encoded_sequences.shape)

print(encoded_sequences[0][:20])
```

```python
np.save(
    "encoded_sequences_len60_v2.npy",
    encoded_sequences
)
```
