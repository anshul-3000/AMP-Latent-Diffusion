```python
import pandas as pd

general = pd.read_excel("../raw/general_amps.xlsx")

# Keep only needed columns
df = general[
    [
        "DRAMP_ID",
        "Sequence",
        "Sequence_Length",
        "Activity",
        "Hemolytic_activity",
        "Target_Organism"
    ]
].copy()

print(df.shape)

# Remove missing sequences
df = df.dropna(subset=["Sequence"])

# Remove duplicates
df = df.drop_duplicates(subset=["Sequence"])

print("After deduplication:", df.shape)

df.head()
```

```python
print(df["Sequence_Length"].describe())
```

```python
invalid = df[
    ~df["Sequence"].str.match(r"^[ACDEFGHIKLMNPQRSTVWY]+$", na=False)
]

print("Invalid sequences:", len(invalid))

invalid.head()
```

```python
VALID_AA = set("ACDEFGHIKLMNPQRSTVWY")

clean_df = df[
    df["Sequence"].apply(
        lambda seq: isinstance(seq, str) and set(seq).issubset(VALID_AA)
    )
].copy()

print("After AA filtering:", clean_df.shape)

clean_df = clean_df[
    (clean_df["Sequence_Length"] >= 5)
    & (clean_df["Sequence_Length"] <= 60)
].copy()

print("After length filtering:", clean_df.shape)

clean_df.reset_index(drop=True, inplace=True)

clean_df.head()
```

```python
hemo = clean_df["Hemolytic_activity"].fillna("")

print("Total samples:", len(hemo))

print(
    hemo.str.contains(
        "hemolytic|hemolysis",
        case=False,
        na=False
    ).sum()
)
```
