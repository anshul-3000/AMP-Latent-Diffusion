```python
import pandas as pd

nonamp = pd.read_csv(
    "../raw/uniprotkb_reviewed_true_AND_length_5_TO_2026_06_03.tsv",
    sep="\t"
)

print(nonamp.shape)
print(nonamp.columns.tolist())
```

```python
print("Before:", nonamp.shape)

nonamp = nonamp.drop_duplicates(subset=["Sequence"])

print("After:", nonamp.shape)
```

```python
amp_df = pd.read_csv(
    "clean_amp_dataset.csv"
)

amp_sequences = set(amp_df["Sequence"])

overlap = nonamp[
    nonamp["Sequence"].isin(amp_sequences)
]

print("Overlap sequences:", len(overlap))
```

```python
import pandas as pd

amp_df = pd.read_csv(
    "clean_amp_dataset.csv"
)

amp_sequences = set(amp_df["Sequence"])

nonamp_clean = nonamp[
    ~nonamp["Sequence"].isin(amp_sequences)
].copy()

print("After overlap removal:", nonamp_clean.shape)
```

```python
print(nonamp_clean["Length"].describe())
```

```python
nonamp_final = nonamp_clean.sample(
    n=8400,
    random_state=42
)

print(nonamp_final.shape)
```

```python
nonamp_final.to_csv(
    "clean_nonamp_dataset.csv",
    index=False
)

print("Saved")
```

```python
amp_df = pd.read_csv(
    "clean_amp_dataset.csv"
)

nonamp_df = pd.read_csv(
    "clean_nonamp_dataset.csv"
)

amp_final = pd.DataFrame({
    "Sequence": amp_df["Sequence"],
    "label": 1
})

nonamp_final = pd.DataFrame({
    "Sequence": nonamp_df["Sequence"],
    "label": 0
})

merged_df = pd.concat(
    [amp_final, nonamp_final],
    ignore_index=True
)

print(merged_df.shape)

print(
    merged_df["label"].value_counts()
)
```

```python
merged_df = merged_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

merged_df.to_csv(
    "amp_nonamp_dataset.csv",
    index=False
)

print(merged_df.shape)

merged_df.head()
```
